```python
import urllib.request
import gzip
import networkx as nx
import pandas as pd
import time
import community.community_louvain as community_louvain # pip install python-louvain


# 1. DATA EXTRACTION 

print("Extracting dataset...")
url = "https://snap.stanford.edu/data/wiki-Vote.txt.gz"
file_name = "wiki-Vote.txt.gz"

# Programmatically downloading the dataset
urllib.request.urlretrieve(url, file_name)

# Instead of downloading from the url, and load the file using pd.load()


# ==========================================
# 2. PREPROCESSING & GRAPH CONSTRUCTION

print("Building the graph...")
# Reading the edgelist, ignoring the metadata comments at the top of the file
edges = pd.read_csv(file_name, sep='\t', comment='#', names=['Source', 'Target'], compression='gzip')

# Building a Directed Graph because a vote from User A to User B is not strictly reciprocal
G = nx.from_pandas_edgelist(edges, source='Source', target='Target', create_using=nx.DiGraph())

# Removing self-loops (users voting for themselves, which skews influence metrics)
G.remove_edges_from(nx.selfloop_edges(G))

# Extracting the Largest Weakly Connected Component to ensure mathematical stability for algorithms
largest_cc = max(nx.weakly_connected_components(G), key=len)
G_core = G.subgraph(largest_cc).copy()

print(f"Graph constructed: {G_core.number_of_nodes()} nodes, {G_core.number_of_edges()} edges.")

# ==========================================
# 3. CALCULATING CENTRALITY METRICS
# ==========================================
print("Calculating Centralities (this may take a minute)...")
start_time = time.time()

# In-Degree (Measures direct votes received)
degree_cent = nx.in_degree_centrality(G_core)

# PageRank (Measures influence based on the influence of voters)
pagerank_cent = nx.pagerank(G_core, alpha=0.85)

# Eigenvector Centrality (Requires a strongly connected graph, often struggles to converge on directed graphs)
try:
    eigen_cent = nx.eigenvector_centrality(G_core, max_iter=1000)
except nx.PowerIterationFailedConvergence:
    eigen_cent = "Failed to converge (common limitation in directed networks without strong connectivity)"

# Betweenness Centrality (Measures who acts as a bridge/gatekeeper in the network)
# We use k=100 (stochastic approximation) to save compute time, as exact calculation is O(V*E)
betweenness_cent = nx.betweenness_centrality(G_core, k=100, seed=42) 

print(f"Metrics calculated in {round(time.time() - start_time, 2)} seconds.")

# ==========================================
# 4. COMMUNITY DETECTION
# ==========================================
print("Detecting Communities...")
# The Louvain algorithm (modularity optimization) requires an undirected graph
G_undirected = G_core.to_undirected()
partition = community_louvain.best_partition(G_undirected)

# Count how many unique communities were found
num_communities = len(set(partition.values()))
print(f"Detected {num_communities} distinct communities.")

# ==========================================
# 5. INFLUENCER STABILITY ACROSS COMMUNITIES
# ==========================================
print("Analyzing Influencer Stability...")

# Get the top 10 influencers globally based on PageRank
top_10_global = sorted(pagerank_cent.items(), key=lambda x: x[1], reverse=True)[:10]

# Group nodes by their detected community
community_dict = {}
for node, comm_id in partition.items():
    if comm_id not in community_dict:
        community_dict[comm_id] = []
    community_dict[comm_id].append(node)

# Check if the global top 10 are distributed across different communities or clustered in one
print("\nTop 10 Global Influencers (PageRank) and their Communities:")
for rank, (node, score) in enumerate(top_10_global, 1):
    comm_id = partition.get(node, "Unknown")
    # Calculate the node's local influence (In-Degree) within its own community
    local_nodes = community_dict[comm_id]
    local_in_degree = sum(1 for src, dst in G_core.in_edges(node) if src in local_nodes)
    
    print(f"{rank}. User {node}: Community {comm_id} | Global PR: {score:.4f} | Local In-Degree: {local_in_degree}")

# Logic for the report: If the top 10 users all belong to different communities, 
# influential users are stable (each community has a distinct leader). 
# If they all belong to Community 0, influence is highly centralized.

# ==========================================
# 6. CENTRALITY COMPARISON OUTPUT
# ==========================================
print("\n--- Centrality Comparison (Top 5 Users) ---")

# In-Degree (Popularity)
top_degree = sorted(degree_cent.items(), key=lambda x: x[1], reverse=True)[:5]
print(f"Top 5 In-Degree:   {[u for u, v in top_degree]}")

# Betweenness (Gatekeepers)
top_bet = sorted(betweenness_cent.items(), key=lambda x: x[1], reverse=True)[:5]
print(f"Top 5 Betweenness: {[u for u, v in top_bet]}")

# Eigenvector (Influence connected to other influencers)
if isinstance(eigen_cent, dict):
    top_eig = sorted(eigen_cent.items(), key=lambda x: x[1], reverse=True)[:5]
    print(f"Top 5 Eigenvector: {[u for u, v in top_eig]}")
else:
    print(f"Eigenvector:       {eigen_cent}")
```
